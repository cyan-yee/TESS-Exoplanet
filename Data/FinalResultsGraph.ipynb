```python
import numpy as np
import pandas as pd
import batman
import matplotlib.pyplot as plt

plt.rcParams.update({'font.size': 12})

def batman_curve(time, rp, a, inc, t0, per, ecc, w, u=[0.3,0.2], limb_dark="quadratic"):
    params = batman.TransitParams()
    params.t0 = t0
    params.per = per
    params.rp = rp
    params.a = a
    params.inc = inc
    params.ecc = ecc
    params.w = w
    params.u = u
    params.limb_dark = limb_dark
    m = batman.TransitModel(params, time)
    return m.light_curve(params)

fig, axes = plt.subplots(2, 2, figsize=(15,10))

data = pd.read_csv('1009_01_TIC107782586_cleaned_lc.csv')
flux= data['flux_normalized'].values
t = data['time_btjd'].values
flux_err = data['flux_err'].values
cond1 = t > 2231.5
cond2 = t < 2232.8
t = t[cond1 & cond2] 
flux = flux[cond1 & cond2]
flux_err = flux_err[cond1 & cond2]

model_best = batman_curve(t, 0.036, 4.55, 87.94, 2232.14, 2.43, 0.81, 268.91)
axes[0,0].scatter(t, flux, s=2, color="black")
axes[0,0].plot(t, model_best, color="C1", lw=2)
axes[0,0].set_xlabel("Time [BTJD]")
axes[0,0].set_ylabel("Relative flux")
axes[0,0].set_title("TIC 107782586")

data = pd.read_csv('1039_01_TIC461867584_cleaned_lc.csv')
flux= data['flux_normalized'].values
t = data['time_btjd'].values
flux_err = data['flux_err'].values
cond1 = t > 2285
cond2 = t < 2288
t = t[cond1 & cond2] 
flux = flux[cond1 & cond2]
flux_err = flux_err[cond1 & cond2]

model_best = batman_curve(t, 0.033, 2.89, 87.93, 2285.8, 1.26, 0.84, 268)
axes[0, 1].scatter(t, flux, s=2, color="black")
axes[0, 1].plot(t, model_best, color="C1", lw=1)
axes[0, 1].set_xlabel("Time [BTJD]")
axes[0, 1].set_ylabel("Relative flux")
axes[0, 1].set_title("TIC 461867584")

data = pd.read_csv('1114_01_TIC409934330_cleaned_lc.csv')
flux= data['flux_normalized'].values
t = data['time_btjd'].values
flux_err = data['flux_err'].values
cond1 = t > 2046.25
cond2 = t < 2047.05
t = t[cond1 & cond2] 
flux = flux[cond1 & cond2]
flux=flux/np.mean(flux)
flux_err = flux_err[cond1 & cond2]

model_best = batman_curve(t, 0.08, 5.67, 88.89, 2046.7, 2.01, 0.61, 248.31)
axes[1, 0].scatter(t, flux, s=2, color="black")
axes[1, 0].plot(t, model_best, color="C1", lw=1)
axes[1, 0].set_xlabel("Time [BTJD]")
axes[1, 0].set_ylabel("Relative flux")
axes[1, 0].set_title("TIC 409934330")

data = pd.read_csv('1149_01_TIC117789567_cleaned_lc.csv')
flux= data['flux_normalized'].values
t = data['time_btjd'].values
flux_err = data['flux_err'].values
cond1 = t > 2801
cond2 = t < 2802.4
t = t[cond1 & cond2] 
flux = flux[cond1 & cond2]
flux_err = flux_err[cond1 & cond2]

model_best = batman_curve(t, 0.07, 5.07, 87.72, 2801.41, 2.21, 0.63, 260.9)
axes[1, 1].scatter(t, flux, s=2, color="black")
axes[1, 1].plot(t, model_best, color="C1", lw=1)
axes[1, 1].set_xlabel("Time [BTJD]")
axes[1, 1].set_ylabel("Relative flux")
axes[1, 1].set_title("TIC 117789567")
plt.tight_layout()
plt.savefig("results1.pdf")
plt.close()
```
